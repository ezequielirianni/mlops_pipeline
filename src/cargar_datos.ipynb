# Carga de Datos

Este notebook simula la carga de datos que normalmente vendría del DWH/Datalake de la empresa como resultado de un proceso de materialización de tablas (ETL). 

Para este proyecto, se utiliza un dataset no productivo de ejemplo en formato `.xlsx`, ubicado en la raíz del repositorio: `Base_de_datos.xlsx`.

El dataset contiene información histórica de créditos otorgados, con el objetivo de predecir el comportamiento de pago de nuevos clientes (variable objetivo: `Pago_atiempo`).

```python
# Importación de librerías
import pandas as pd
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

```python
# Definición de rutas
# El notebook vive en src/, la base de datos está un nivel arriba, en la raíz del repo
DATA_PATH = Path("..") / "Base_de_datos.xlsx"

print(f"Buscando archivo en: {DATA_PATH.resolve()}")
print(f"¿Existe el archivo?: {DATA_PATH.exists()}")
```

```python
# Carga del dataset
df = pd.read_excel(DATA_PATH)

print(f"Dataset cargado correctamente.")
print(f"Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
```

## Exploración rápida (sanity check)

Antes de pasar al notebook de EDA, hacemos una validación mínima: que el archivo cargó bien, que las columnas esperadas están presentes y que no está vacío.

```python
# Vista previa de los primeros registros
df.head()
```

```python
# Nombres de columnas y tipos de dato originales (tal cual vienen del Excel)
df.info()
```

```python
# Validación mínima de integridad
assert df.shape[0] > 0, "El dataset está vacío"
assert "Pago_atiempo" in df.columns, "No se encontró la variable objetivo 'Pago_atiempo'"

print("Validación de carga exitosa. El dataset está listo para el análisis exploratorio (EDA).")
```
