# Carga de Datos

Este notebook simula la carga de datos que normalmente vendría del DWH/Datalake de la empresa como resultado de un proceso de materialización de tablas (ETL).

Para este proyecto, se utiliza un dataset no productivo de ejemplo en formato `.xlsx`, ubicado en la raíz del repositorio: `Base_de_datos.xlsx`.

El dataset contiene información histórica de créditos otorgados, con el objetivo de predecir el comportamiento de pago de nuevos clientes (variable objetivo: `Pago_atiempo`).

```python
# Importación de librerías
import pandas as pd
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

## Resolución de rutas

Se evita depender de rutas relativas frágiles como `Path("..")`, cuyo resultado cambia según desde dónde se lance el kernel (VS Code, Jupyter Lab). En su lugar, se busca la raíz del repositorio subiendo desde el directorio actual hasta encontrar un archivo "marca" (`requirements.txt`), que por la estructura fija del proyecto siempre vive en la raíz.

```python
def find_repo_root(marker: str = "requirements.txt", start: Path = None) -> Path:
    """Busca hacia arriba desde 'start' (o el directorio actual) hasta encontrar
    la carpeta que contiene 'marker'. Es independiente de desde donde se lance el kernel."""
    current = (start or Path.cwd()).resolve()
    for parent in [current, *current.parents]:
        if (parent / marker).exists():
            return parent
    raise FileNotFoundError(
        f"No se encontro '{marker}' en ningun directorio padre a partir de {current}"
    )

REPO_ROOT = find_repo_root()
DATA_PATH = REPO_ROOT / "Base_de_datos.xlsx"

print(f"Raiz del repo detectada: {REPO_ROOT}")
print(f"Buscando archivo en: {DATA_PATH}")
print(f"Existe el archivo?: {DATA_PATH.exists()}")
```

```python
# Carga del dataset
df = pd.read_excel(DATA_PATH)

print(f"Dataset cargado correctamente.")
print(f"Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
```

### Validación de columnas esperadas

En vez de asumir que "algunas" columnas están presentes, se compara explícitamente contra el set completo de columnas esperado según el diccionario de datos del proyecto.

```python
EXPECTED_COLUMNS = {
    'tipo_credito', 'fecha_prestamo', 'capital_prestado', 'plazo_meses', 'edad_cliente',
    'tipo_laboral', 'salario_cliente', 'total_otros_prestamos', 'cuota_pactada', 'puntaje',
    'puntaje_datacredito', 'cant_creditosvigentes', 'huella_consulta', 'saldo_mora',
    'saldo_total', 'saldo_principal', 'saldo_mora_codeudor', 'creditos_sectorFinanciero',
    'creditos_sectorCooperativo', 'creditos_sectorReal', 'promedio_ingresos_datacredito',
    'tendencia_ingresos', 'Pago_atiempo'
}

columnas_faltantes = EXPECTED_COLUMNS - set(df.columns)
columnas_inesperadas = set(df.columns) - EXPECTED_COLUMNS

print(f"Columnas faltantes: {columnas_faltantes if columnas_faltantes else 'ninguna'}")
print(f"Columnas inesperadas (no documentadas): {columnas_inesperadas if columnas_inesperadas else 'ninguna'}")

assert not columnas_faltantes, f"Faltan columnas esperadas: {columnas_faltantes}"
print("Validacion de columnas esperadas superada.")
```

### Duplicados y estado del target

```python
duplicados = df.duplicated().sum()
print(f"Filas duplicadas: {duplicados}")

nulos_target = df['Pago_atiempo'].isnull().sum()
print(f"Nulos en la variable objetivo (Pago_atiempo): {nulos_target}")
```

```python
# Validación mínima de integridad
assert df.shape[0] > 0, "El dataset esta vacio"
assert "Pago_atiempo" in df.columns, "No se encontro la variable objetivo 'Pago_atiempo'"
assert nulos_target == 0, "La variable objetivo tiene valores nulos"

print("Validacion de carga exitosa. El dataset esta listo para el analisis exploratorio (EDA).")
```
